```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
df_datatest = pd.read_csv('data/AIAAIC Repository - Incidents.csv')
df_datatest.head()
```

```python
anni_specificati = ['2004','2005','2006','2007', '2008','2009','2010','2011', '2012','2013','2014','2015','2016','2017', '2018', '2019', '2020','2021', '2022', '2023', '2024', '2025']
df_datatest['Occurred'] = df_datatest['Occurred'].apply(lambda x: x if x in anni_specificati else 'Altro')
print(df_datatest['Occurred'].isna().sum())
# Raggruppa i dati per anno
df_grouped = df_datatest.groupby('Occurred').size().reset_index(name='Count')
df_grouped
```

```python
plt.figure(figsize=(10,5))
plt.plot(df_grouped["Occurred"][0:-1], df_grouped["Count"][0:-1], marker="o", linestyle="-", color="b", label="Incidenti")

# Aggiunta di titoli ed etichette
plt.xlabel("Anno")
plt.ylabel("Numero di Incidenti")
plt.title("Evoluzione del Numero di Incidenti (2008-marzo 2025)")
plt.xticks(rotation=45)  # Ruota le etichette sull'asse X per leggibilità
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

# Mostra il grafico
plt.show()
```

```python
# Dati sugli incidenti
incidenti_labels = ["Disinformazione pubblica", "Diffamazione", "Frodi", "Sfruttamento", "Furto di identità"]
incidenti_valori = [60, 15, 10, 8, 7]

# Dati sui target presi di mira
target_labels = ["Figure politiche e organizzazioni", "Settore dei media", "Industria", "Settore privato"]
target_valori = [54, 28, 9, 8]

# Creazione della figura con due grafici a torta
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

# Grafico a torta per gli incidenti
ax[0].pie(incidenti_valori, labels=incidenti_labels, autopct='%1.1f%%', colors=["#ff9999", "#66b3ff", "#99ff99", "#ffcc99", "#c2c2f0"], startangle=140)
ax[0].set_title("Tipologia di Incidenti")

# Grafico a torta per i target
ax[1].pie(target_valori, labels=target_labels, autopct='%1.1f%%', colors=["#ff6666", "#ffb366", "#66ff66", "#66b3ff"], startangle=140)
ax[1].set_title("Target Presi di Mira")

# Mostra il grafico
plt.show()
```
